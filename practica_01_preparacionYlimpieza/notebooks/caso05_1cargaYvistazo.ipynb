Este cuaderno sirve para conocer el CSV original antes de limpiarlo. Revisa su tamaño, tipos de datos y estadísticas, y busca valores ausentes tanto explícitos como escritos de otra forma. También muestra espacios sobrantes, categorías, frecuencias y duplicados. Es una revisión descriptiva: aquí no se modifican los datos.


## Carga de datos

Se importan las librerías y se lee el CSV original.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("../../data/raw/sonrievalencia_citas.csv", sep=";")

print("Dataset cargado correctamente")
```

## Primer vistazo y dimensiones

Se muestra una muestra de filas, el tamaño del conjunto y sus tipos de datos.

```python
display(df.head())

print("Dimensiones del dataset:", df.shape)
print("Número de filas:", df.shape[0])
print("Número de columnas:", df.shape[1])

df.info()
```

## Estadísticas y valores ausentes

El resumen descriptivo y los recuentos permiten revisar la distribución y los datos que pandas reconoce como nulos.

```python
display(df.describe(include="all"))

print("Valores nulos por columna:")
display(df.isnull().sum())

print("Porcentaje de valores nulos:")
display((df.isnull().mean() * 100).round(2))
```

## Nulos escritos como texto y espacios sobrantes

Se buscan marcadores frecuentes de datos ausentes y espacios al inicio o al final de las columnas de texto.

```python
valores_nulos = [
    "-", "?", "N/A", "NA", "NULL",
    "NONE", "DESCONOCIDO", "NO DISPONIBLE"
]

print("BÚSQUEDA DE POSIBLES NULOS DISFRAZADOS")
for columna in df.select_dtypes(include="object").columns:
    valores = df[columna].astype("string").str.strip().str.upper()
    encontrados = valores.isin(valores_nulos).sum()

    if encontrados > 0:
        print(f"{columna}: {encontrados} posibles nulos disfrazados")

print("BÚSQUEDA DE ESPACIOS AL PRINCIPIO O FINAL")
for columna in df.select_dtypes(include="object").columns:
    espacios = df[columna].astype("string").str.contains(
        r"^\s|\s$", na=False
    ).sum()

    if espacios > 0:
        print(f"{columna}: {espacios} valores con espacios")
```

## Categorías y frecuencias

Se listan los valores distintos y la frecuencia de cada uno, incluyendo los valores nulos.

```python
print("VALORES ÚNICOS DE CADA COLUMNA")
for columna in df.columns:
    print(f"\n--- {columna} ---")
    print(df[columna].unique())

print("FRECUENCIA DE VALORES POR COLUMNA")
for columna in df.columns:
    print(f"\n--- {columna} ---")
    print(df[columna].value_counts(dropna=False))
```

## Resumen

Se recoge el tamaño del conjunto, los nulos y los duplicados exactos.

```python
print("\n========== RESUMEN DEL DATASET ==========")
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])
print("Total de valores nulos:", df.isnull().sum().sum())
print("Columnas con valores nulos:", df.isnull().gt(0).sum())
print("Duplicados:", df.duplicated().sum())
print("=========================================")
```
