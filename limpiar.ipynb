# Limpieza de datos: `edad`, `recordatorio_sms` y `urgencia`

```python
import pandas as pd
import numpy as np

# 1. Cargar el dataset
df = pd.read_csv("sonrievalencia_citas.csv", sep=";")

# ---------------------------------------------------------
# 2. Limpieza de 'edad': corregir atípicos (>100) e imputar con la mediana
# ---------------------------------------------------------
df.loc[df['edad'] > 100, 'edad'] = np.nan
mediana_edad = df['edad'].median()
df['edad'] = df['edad'].fillna(mediana_edad)

# ---------------------------------------------------------
# 3. Limpieza de 'recordatorio_sms': asumir 'No' en valores nulos
# ---------------------------------------------------------
df['recordatorio_sms'] = df['recordatorio_sms'].fillna('No')

# ---------------------------------------------------------
# 4. Limpieza de 'urgencia': imputar con la moda (opción más frecuente)
# ---------------------------------------------------------
moda_urgencia = df['urgencia'].mode()[0]
df['urgencia'] = df['urgencia'].fillna(moda_urgencia)

# ---------------------------------------------------------
# 5. Comprobación de nulos restantes
# ---------------------------------------------------------
print("Valores nulos tras la primera limpieza:")
print(df[['edad', 'recordatorio_sms', 'urgencia']].isnull().sum())

print(f"\nMediana aplicada a edad: {mediana_edad}")
print(f"Moda aplicada a urgencia: {moda_urgencia}")
```

# Limpieza de datos: `seguro`

```python
# ---------------------------------------------------------
# Limpieza de la columna 'seguro'
# ---------------------------------------------------------
# 1. Homogeneizar texto: eliminar espacios en blanco sobrantes y unificar a tipo título/capitalizado
df['seguro'] = df['seguro'].astype(str).str.strip().str.capitalize()

# 2. Convertir el nulo disfrazado ('Desconocido') a NaN
df.loc[df['seguro'] == 'Desconocido', 'seguro'] = np.nan

# 3. Calcular la moda (el valor más frecuente) entre los seguros válidos
moda_seguro = df['seguro'].mode()[0]
print(f"Moda aplicada a 'seguro': {moda_seguro}")

# 4. Imputar los nulos con la moda
df['seguro'] = df['seguro'].fillna(moda_seguro)

# 5. Comprobación de resultados finales
print(f"Nulos restantes en 'seguro': {df['seguro'].isnull().sum()}")
print("\nDistribución final de categorías en 'seguro':")
print(df['seguro'].value_counts())
```

# Guardar el dataset limpio en un nuevo archivo CSV

```python
# Guardar el DataFrame con los datos limpios en un nuevo CSV
archivo_salida = "sonrievalencia_citas_limpio.csv"
df.to_csv(archivo_salida, sep=";", index=False, encoding="utf-8")

print(f"Dataset limpio guardado exitosamente en: '{archivo_salida}'")
print(f"Dimensiones del archivo guardado: {df.shape}")
```
